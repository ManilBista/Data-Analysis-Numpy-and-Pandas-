NumPy is particularly useful because it allows efficient vectorized numerical operations.

```python
## for example
numbers = np.array([10, 20, 30, 40, 50])

print(numbers * 2)
```

```python
## with normal python code
numbers = [10, 20, 30]

print(numbers * 2)
```

```python
# 2d array
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(a)
```

```python
print(a.ndim)   ##checks the dimensions of the array
```

```python
## for rows and colums. value of x defines no. of rows and y defines columns
print(a.shape)
```

```python
## total no. of elements
print(a.size)
```

```python
## for datatypes
print(a.dtype)
```

```python
a = np.array([10, 20, 30, 40])

print("Dimensions:", a.ndim)
print("Shape:", a.shape)
print("Size:", a.size)
print("Data Type:", a.dtype)
```

```python
## special arrays
a = np.zeros(5)    # creates array of 5 0s

print(a)
```

```python
# 0 array for 2D
a = np.zeros((2, 3))

print(a)
```

```python
# 1s array
a = np.ones(5)

print(a)
```

```python
# for 2D of 1s array
a = np.ones((2,3))

print(a)
```

```python
# 3D array of ones
a = np.ones((2, 3, 4)) # 2 layers, 3 rows, 4 columns

print(a)
```

```python
## full array (creates array full of 10s 5times)
a = np.full(5, 10)

print(a)
```
