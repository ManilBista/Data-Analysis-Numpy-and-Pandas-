**1. Series & DataFrame Basics**

Concept: A Series is a single labeled column of data: like one column in a spreadsheet, with row labels (an index) attached. A DataFrame is a collection of Series sharing the same index, the full spreadsheet, not just one column.

```python
import pandas as pd
```

```python
s = pd.Series([10, 20, 30])
print(s)
df = pd.DataFrame({'name': ['Ram', 'Sita'], 'age': [20, 22]})
print(df)
```

Key attributes:

Attribute/Method and Meaning

.index	(Row labels)

.columns	(Column names (DataFrame only))

.values	(Underlying data as a NumPy array)

.shape	   (rows, columns)

.dtypes	   (Data type of each column)

.head(n), .tail(n)	(First/last n rows)

.info()	  (Summary: columns, types, non-null counts)

**2. Reading/Creating Data (dict/list/CSV)**

Concept: DataFrames can be built directly from Python structures already in memory, or read from external files, same object, different source.

```python
# from dict
df1 = pd.DataFrame({'a': [1,2], 'b': [3,4]})

# from list of lists (need column names)
df2 = pd.DataFrame([[1,3],[2,4]], columns=['a','b'])

# from CSV
df3 = pd.read_csv('data.csv')
```

**Key operations:**

**Function	Purpose**

pd.DataFrame(dict)	---Build from dict of lists (keys = columns)

pd.DataFrame(list, columns=)---	Build from list of rows

pd.read_csv(path)	---Load DataFrame from CSV file

pd.read_csv(path, index_col=0)	---Use first column as index

**3. Indexing & Selection (loc/iloc)**

**Concept:**  .loc selects by label (the named index/column), .iloc selects by integer position — like the difference between asking for "the row named Ram" versus "the 3rd row, whatever its name is."

```python
df = pd.DataFrame({'name': ['Ram','Sita','Hari'], 'age': [20,22,25]}, index=['a','b','c'])
```

```python
df.loc['a']          # row labeled 'a' (by name)
```

```python
df.loc['a', 'age']   # single value by label
```

```python
df.iloc[0]           # first row (by position)
```

```python
df.iloc[0, 1]        # first row, second column, by position
```

```python
df['name']           # single column (returns a Series)
```

```python
df[['name','age']]   # multiple columns (returns a DataFrame)
```

**Key patterns:**

**Pattern	Meaning**

df['col']---	Select one column

df[['col1','col2']]---	Select multiple columns

df.loc[label]	---Select row(s) by label

df.loc[label, col]---	Select cell by label

df.iloc[pos]	---Select row(s) by integer position

df.iloc[pos, pos]---	Select cell by position
