**4. Filtering & Boolean Masks**

**Concept:** A boolean mask is a Series of True/False values, one per row, used to keep only the rows that satisfy a condition: like a sieve that only lets matching rows through.

```python
import pandas as pd
```

```python
df = pd.DataFrame({'name':['Ram','Sita','Hari'],
                   'score':[70,85,60]})
```

```python
mask = df['score'] > 65
print(mask)                   # True/False per row
```

```python
df[df['score'] > 65]                      # rows where score > 65
```

```python
df[(df['score'] > 60) & (df['score'] < 90)]  # multiple conditions (& and |, not "and"/"or")
```

**Pattern	and Meaning**

df[df['col'] > x]	---Filter rows by condition
& | ~	AND / OR / NOT for combining conditions

df['col'].isin([...])	---Filter by membership in a list

df['col'].between(a, b)	---Filter by range

**5. Handling Missing Data**

Concept: Real datasets have gaps, represented as NaN. Pandas gives tools to detect, remove, or fill these gaps rather than crashing on them.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({'a': [1, np.nan, 3], 'b': [4, 5, np.nan]})
```

```python
df.isna()             # True/False mask of missing values
```

```python
df.isna().sum()       # count of missing values per column
```

```python
df.dropna()           # drop rows with any NaN
```

```python
df.fillna(0)          # replace NaN with 0
```

```python
df['a'].fillna(df['a'].mean())   # fill with column mean
```

**Key operations:**

**Function	Purpose**

.isna() / .isnull()	---Detect missing values

.dropna()	---Remove rows/columns with missing values

.fillna(value)	---Replace missing values

.fillna(method='ffill')	---Fill using previous valid value
