**Array Operations & Broadcasting**

Concept: Operations apply element-wise by default, no loops needed.

 Broadcasting is NumPy's rule for handling operations between arrays of different shapes by "stretching" the smaller one, like automatically applying a single tax rate to every item in a cart without writing a loop.

```python
import numpy as np
```

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

a + b        #  element-wise


```

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
a * b            # element-wise multiply (not matrix mult)
```

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
a * 2
```

```python
m = np.array([[1,2,3],[4,5,6]])
m + a        # adds [1,2,3] to every row -> broadcasting
```

**Broadcasting rule** (compare shapes right to left):

dimensions are compatible if they are equal, or one of them is 1.

Key operations:

Operation	Example

'+' '-' '*' '/'	(Element-wise arithmetic)

**	 (Element-wise power)

np.sqrt(a)	(Element-wise square root)

np.exp(a)	(Element-wise exponential)

a @ b or np.dot(a,b)	(Matrix/dot product (NOT *))

**Reshaping, Stacking, Splitting**

Concept: Same data, different "view" of its shape. Total element count must stay constant when reshaping.

```python
a = np.arange(12)
a.reshape(3, 4)      # reshape into 3 rows, 4 cols
a.reshape(3, -1)     # -1 = "figure this dimension out automatically"

a.flatten()          # collapse to 1D (returns a copy)
a.T                  # transpose (swap axes)

np.vstack([a, a])    # stack arrays vertically
np.hstack([a, a])    # stack arrays horizontally
np.split(a, 3)       # split into 3 equal parts
```

Key operations:

Operation	Purpose

.reshape(shape)	Change shape without changing data

.flatten() / .ravel()	Flatten to 1D

.T	Transpose

np.vstack, np.hstack	Stack arrays vertically/horizontally

np.concatenate([a,b], axis=)	Join arrays along an axis

np.split(a, n)	Split into n equal parts

**Aggregation & Statistics**

Concept: Functions that collapse an array into a summary value-- total, average, extremes: optionally along a specific axis (row-wise or column-wise) instead of the whole array.

```python
a = np.array([[1,2,3],[4,5,6]])
```

```python
a.sum()         #  sum of all elements
```

```python
a.sum(axis=0)   #  sum down each column
```

```python
a.sum(axis=1)   # sum across each row
```

```python
a.mean()        # average
```

```python
a.std()         # standard deviation
```

```python
a.min(), a.max()
```

```python
a.argmax()      # index of the max value (flattened)
```

**Axis rule of thumb**:

 axis=0 collapses rows (result has one value per column).

 axis=1 collapses columns (one value per row).

Key operations:

Function	Purpose:

.sum(), .mean(), .std(), .var()

Basic statistics:

.min(), .max()

Extremes:

.argmin(), .argmax()

Index of extremes:

np.median(a)

Median:

.cumsum()

Running cumulative sum
