**Linear Algebra Essentials**

Concept: NumPy's linalg module covers the matrix math behind ML and data science: determinants, inverses, eigenvalues.

```python
import numpy as np
```

```python
m = np.array([[1,2],[3,4]])
```

```python
np.linalg.det(m)         # determinant
```

```python
np.linalg.inv(m)         # inverse matrix
```

```python
np.linalg.eig(m)         # eigenvalues, eigenvectors
```

```python
np.linalg.solve(m, [1,2])  # solve system m @ x = [1,2]
```

**The idea:**

m @ x = b is matrix notation for a system of equations. np.linalg.solve(m, b) finds the vector x that makes this true, without you manually doing elimination.

Concretely, for m = [[1,2],[3,4]] and b = [1,2]:

m @ x = b unpacks into two equations (where x = [x0, x1]):

1* x0 + 2* x1 = 1

3* x0 + 4* x1 = 2

np.linalg.solve(m, [1,2]) solves both equations at once and returns x0, x1.

example:

m = np.array([[1,2],[3,4]])

b = np.array([1,2])

x = np.linalg.solve(m, b)

print(x)        --[0.  0.5]

# verify: m @ x should equal b
print(m @ x)      -- [1. 2.]

**Random Module**

Concept: Used constantly for simulations, sampling, and shuffling data. Set a seed for reproducibility: same seed always gives the same "random" output.

```python
np.random.seed(42)          # reproducibility
```

```python
np.random.rand(3)           # 3 uniform floats [0,1)
```

```python
np.random.randn(3)          # 3 samples from standard normal distribution
```

```python
np.random.randint(1, 100, 5)  # 5 random ints
```

```python
np.random.choice([1,2,3], size=2, replace=False)  # random sample, no repeats
```

```python
np.random.shuffle(array)      # shuffle in place
```

**File I/O**

Concept: Save/load arrays to disk without converting to Python lists first.

```python
a = np.array([1,2,3])

```

```python
np.save('arr.npy', a)          # save single array (binary)
```

```python
b = np.load('arr.npy')         # load it back

```

```python
np.savetxt('arr.csv', a, delimiter=',')   # save as text/CSV
c = np.loadtxt('arr.csv', delimiter=',')  # load from CSV
```
